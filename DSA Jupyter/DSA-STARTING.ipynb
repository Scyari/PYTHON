1. Finding lowest number from an array 
   

```python
array = [5,10,12,2,4]
minValue = array [0]   #so minValue = 5 cz of index

for i in array:        #i is reffering each data from array
    if i < minValue:   # condition to check each value is less than from minValue or array[0]
      minValue = i     # picking the smallest number

print (f"Lowest number is: {minValue}")
```
