3. sum the numbers of 5 to 1

```python
sum = 0
i = 5
while i > 0:
    sum+=i
    i-=1
print (f"sum the numbers of 5 to 1 is : {sum}") 
```

4. FIZZ BUZZ PROBLEM

```python
i =  1
n = eval (input("Enter the number"))

while i < n:
    if i % 3 == 0 and  i % 5 == 0:
        print ("FizzBuzz")
    elif i % 3 == 0:
        print ("Fizz")
    elif i % 5 == 0:
        print ("Buzz")
    else :
        print (i)
    i+=1
```

8. Print sum of Odd number and Even Number 

```python
i = 1
n = 5
Odd_sum = 0
while i <= n:
    Odd_sum += i
    i+=2
print (f"Sum of odd number is: {Odd_sum}")

j = 2
even_sum = 0
while j <= n:
    even_sum += j
    j+=2
print (f"Sum of even number is: {even_sum}")


```

9. Product of 1 to n (factorial)

```python
i = 1
n = eval (input("Enter the number"))
prod = 1
while i <= n:
    prod *= i
    i+=1
print (f"factorial of {n} : {prod}")
```
