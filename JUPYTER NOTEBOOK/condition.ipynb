1. PASS OR FAIL AVERAGE

```python
phy = eval (input("Enter Number:"))
chem = eval (input("Enter Number:"))
math = eval (input("Enter Number:"))
average = (phy + chem + math) / 3
if phy >=80:
    print("Grade: B, Pass")
else:
    print("Grade: D, Fail")
if chem >=80:
    print("Grade: B, Pass")
else:
    print("Grade: D, Fail")
if math >=80:
    print("Grade: B, Pass")
else:
    print("Grade: D, Fail")
if average >=80:
    print("Grade: A, Pass")
else:
    print("Grade: D, Fail")
```
