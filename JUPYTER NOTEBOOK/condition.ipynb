1. PASS OR FAIL AVERAGE

```python
phy = eval (input("Enter Number:"))
chem = eval (input("Enter Number:"))
math = eval (input("Enter Number:"))
average = (phy + chem + math) / 3
if phy >=80:
    print("Grade: B, Pass")
else:
    print("Grade: D, Fail")
if chem >=80:
    print("Grade: B, Pass")
else:
    print("Grade: D, Fail")
if math >=80:
    print("Grade: B, Pass")
else:
    print("Grade: D, Fail")
if average >=80:
    print("Grade: A, Pass")
else:
    print("Grade: D, Fail")
```

4. ATM WITH-DRAWL SYSTEM

```python
Balance = 2000
CorrectPin = 2026
EnteredPin = eval(input("Enter Pin:"))

if CorrectPin == EnteredPin:
    Withdraw = eval(input("Enter Withdrawl amount:"))       #asking for Withdraw

    if Withdraw <= 0:
      print ("Amount must be a positive number")

    elif Withdraw > Balance:
      print ("Insufficient Balance")

    elif (Balance - Withdraw) < 500:                    
      print ("Minimum Account Balance Must be 500")

    else:
      Balance -= Withdraw
      print ("WithDrawl Successful")
      print ("WithDrawl Amount", Withdraw)
      print ("Remaining Balance", Balance)

else:
     print ("Wrong Pin")



    
```
