4. ATM WITH-DRAWL SYSTEM

```python
Balance = 2000
CorrectPin = 2026
EnteredPin = eval(input("Enter Pin:"))

if CorrectPin == EnteredPin:                                #another Branch
    Withdraw = eval(input("Enter Withdrawl amount:"))       #asking for Withdraw

    if Withdraw <= 0:                                        #Think this like a tree (DSA) one Branch
      print ("Amount must be a positive number")

    elif Withdraw > Balance:
      print ("Insufficient Balance")

    elif (Balance - Withdraw) < 500:                    
      print ("Minimum Account Balance Must be 500")

    else:
      Balance -= Withdraw
      print ("WithDrawl Successful")
      print ("WithDrawl Amount", Withdraw)
      print ("Remaining Balance", Balance)

else:
     print ("Wrong Pin")                                   #another Branch



# Diagram be like -->

          #           PIN correct?
          #          /            \
          #       NO                YES
          #       |                  |
          #  Wrong PIN        Enter withdrawal
          #                         |
          #                  Amount <= 0?
          #                  /         \
          #                YES          NO
          #                |             |
          #             Invalid       Amount > Balance?
          #                             /        \
          #                           YES         NO
          #                           |            |
          #                     Insufficient    Balance < 500?
          #                                        /       \
          #                                      YES        NO
          #                                      |           |
          #                                   Reject      Withdraw   




    
```
