6. find the compound interest!

```python
principle_amount = eval (input("enter principle amount:"))
rate = eval (input("enter rate:"))
time = eval (input("enter time:"))
si = (principle_amount * rate * time) / 100          #rule of simple interest    
ci = principle_amount*(1 + rate / 100) **time        #rule of compound interest
print (f"Simple Interest is: {si}")               
print (f"Compound Interest is: {ci}")             
```

7. find discounted price!

```python
product_price = eval (input("enter product price:"))
Discount_price = eval (input("enter Discount price:"))
discounted_price = (product_price * Discount_price) / 100   
price_after_discount = product_price - discounted_price     
print (f"Discounted price is: {discounted_price}") 
print (f"price after discount is:{price_after_discount}")
```
